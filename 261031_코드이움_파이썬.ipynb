```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

driver = webdriver.Chrome()
driver.maximize_window()

driver.get('https://www.naver.com')

time.sleep(3)

search = WebDriverWait(driver,10).until(
    EC.element_to_be_clickable((By.ID, 'query'))
)

search.send_keys('유민호')

search.send_keys(Keys.ENTER)

image_tab = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable(
        (By.XPATH, "//a[contains(normalize-space(.), '이미지')]")
    )
)

image_tab.click()

time.sleep(3)

driver.quit()
```

```python

from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup
import time

driver = webdriver.Chrome()
driver.maximize_window()

driver.get('https://www.naver.com')

time.sleep(3)

search = WebDriverWait(driver, 10).until(
    EC.element_to_be_clickable((By.ID, 'query'))
)

search.send_keys('이상해씨')
search.send_keys(Keys.ENTER)

WebDriverWait(driver, 10).until(
    EC.presence_of_element_located((By.TAG_NAME, "body"))
)

time.sleep(3)

html = driver.page_source

# HTML 파싱
soup = BeautifulSoup(html, 'html.parser')

print("페이지 제목:", soup.title.get_text(strip=True) if soup.title else "없음")
print("\n===== 검색 결과 텍스트 =====")

content = soup.select_one('#main_pack')

if content:
    for tag in content.select('script, style'):
        tag.decompose()

    text = content.get_text(separator="\n", strip=True)
    print(text[:5000])

    links = content.select("a[href]")

    count = 0

    for link in links:
        title = link.get_text(strip=True)
        url = link.get("href")

        if title and url and url.startswith("http"):
            print("제목:", title)
            print("주소:", url)
            print("-------------------------")

            count += 1

        if count == 10:
            break
else:
    print("검색 결과 영역을 찾을 수 없습니다.")

driver.quit()
```
